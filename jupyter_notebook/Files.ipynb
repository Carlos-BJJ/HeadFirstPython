```python
import os
import webbrowser

import swimclub
```

```python
swimclub.read_swim_data("Abi-10-50m-Free.txt")
```

```python
swim_files = os.listdir(swimclub.FOLDER)
```

```python
print(swim_files)
```

```python
swim_files.remove('.DS_Store')
#Removendo o dado do arquivo que não é para estar ali ('.DS_Store')
```

```python
swimmers = {}
for file in swim_files:
    name, *__ = swimclub.read_swim_data(file)
    if name not in swimmers:
        swimmers[name] = []
    swimmers[name].append(file)
```

```python
for fname in swimmers["Abi"]:
    webbrowser.open("file://" + os.path.realpath(swimclub.produce_bar_chart(fname)))
```
